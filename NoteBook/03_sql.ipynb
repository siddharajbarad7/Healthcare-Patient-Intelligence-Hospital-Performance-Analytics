```python
# medical conndition analysis 
query = """
SELECT
    medical_condition,
    COUNT(*) AS admissions,
    ROUND(SUM(billing_amount), 2) AS total_billing,
    ROUND(AVG(billing_amount), 2) AS average_billing,
    ROUND(AVG(length_of_stay), 2) AS average_length_of_stay
FROM healthcare_records
GROUP BY medical_condition
ORDER BY admissions DESC
"""

condition_analysis = pd.read_sql(query, engine)

condition_analysis
```

```python
query = """
SELECT
    hospital,
    COUNT(*) AS admissions,
    ROUND(SUM(billing_amount), 2) AS total_billing,
    ROUND(AVG(billing_amount), 2) AS average_billing,
    ROUND(AVG(length_of_stay), 2) AS average_length_of_stay
FROM healthcare_records
GROUP BY hospital
ORDER BY total_billing DESC
LIMIT 20
"""

hospital_analysis = pd.read_sql(query, engine)

hospital_analysis
```

```python
query = """
SELECT
    YEAR(date_of_admission) AS admission_year,
    MONTH(date_of_admission) AS admission_month,
    COUNT(*) AS admissions,
    ROUND(SUM(billing_amount), 2) AS total_billing,
    ROUND(AVG(billing_amount), 2) AS average_billing
FROM healthcare_records
GROUP BY
    YEAR(date_of_admission),
    MONTH(date_of_admission)
ORDER BY
    admission_year,
    admission_month
"""

monthly_analysis = pd.read_sql(query, engine)

monthly_analysis
```

```python
# month-over-month growth analysis
query = """
SELECT
    m.month,
    ROUND(m.total_billing, 2) AS total_billing,
    ROUND(p.total_billing, 2) AS previous_month_billing,
    ROUND(
        ((m.total_billing - COALESCE(p.total_billing, 0))
        / NULLIF(COALESCE(p.total_billing, 0), 0)) * 100,
        2
    ) AS mom_growth_pct
FROM (
    SELECT
        DATE_FORMAT(date_of_admission, '%Y-%m') AS month,
        SUM(billing_amount) AS total_billing
    FROM healthcare_records
    GROUP BY DATE_FORMAT(date_of_admission, '%Y-%m')
) AS m
LEFT JOIN (
    SELECT
        DATE_FORMAT(DATE_SUB(date_of_admission, INTERVAL 1 MONTH), '%Y-%m') AS month,
        SUM(billing_amount) AS total_billing
    FROM healthcare_records
    GROUP BY DATE_FORMAT(DATE_SUB(date_of_admission, INTERVAL 1 MONTH), '%Y-%m')
) AS p
    ON p.month = m.month
ORDER BY m.month
"""

try:
    mom_analysis = pd.read_sql(query, engine)
    mom_analysis
except Exception as e:
    print("Month-over-month analysis failed ❌")
    print(e)
```
