```python
# Billing Analysis
df["billing_amount"].describe()
```

```python
avg_bill = df["billing_amount"].mean()

print(f"Average Billing Amount: ${avg_bill:,.2f}")
```

```python
# total billing 
total_billing = df["billing_amount"].sum()

print(f"Total Billing Amount: ${total_billing:,.2f}")
```

```python
# billing by medical condition
billing_condition = (
    df.groupby("medical_condition")
      .agg(
          total_billing=("billing_amount", "sum"),
          avg_billing=("billing_amount", "mean"),
          patients=("name", "count")
      )
      .sort_values("total_billing", ascending=False)
)

billing_condition
```

```python
# insure provide    analysis
insurance_analysis = (
    df.groupby("insurance_provider")
      .agg(
          patients=("name", "count"),
          total_billing=("billing_amount", "sum"),
          avg_billing=("billing_amount", "mean")
      )
      .sort_values("total_billing", ascending=False)
)

insurance_analysis
```

```python

# Time-Series Analysis
monthly_admissions = (
    df.groupby(
        pd.to_datetime(df["date_of_admission"], errors="coerce").dt.to_period("M")
    )
    .size()
)

monthly_admissions
```

```python
# Monthly billing
monthly_billing = (
    df.groupby(
        pd.to_datetime(df["date_of_admission"], errors="coerce").dt.to_period("M")
    )["billing_amount"]
    .sum()
)

monthly_billing
```

```python
plt.figure(figsize=(14, 5))

monthly_billing.plot()

plt.title("Monthly Healthcare Billing Trend")
plt.xlabel("Month")
plt.ylabel("Total Billing")
plt.xticks(rotation=45)
plt.show()
```

```python
# correlation analysis
numeric_cols = [
    "age",
    "billing_amount",
    "room_number",
    "length_of_stay"
]

corr = df[numeric_cols].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f"
)

plt.title("Healthcare Numeric Variable Correlation")
plt.show()
```

```python
# outliner Analysis
Q1 = df["billing_amount"].quantile(0.25)
Q3 = df["billing_amount"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

billing_outliers = df[
    (df["billing_amount"] < lower) |
    (df["billing_amount"] > upper)
]

print("Billing outliers:", len(billing_outliers))
```
