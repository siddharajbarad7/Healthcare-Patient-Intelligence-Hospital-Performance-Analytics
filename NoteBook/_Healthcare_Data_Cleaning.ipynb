```python
missing = df.isnull().sum()

missing_df = pd.DataFrame({
    "Missing Values": missing,
    "Missing %": (missing / len(df) * 100).round(2)
})

missing_df.sort_values("Missing Values", ascending=False)
```

```python
# check it duplicate rows
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count) 
```

```python
df = df.drop_duplicates().copy()

print("Rows after removing duplicates:", len(df))
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

df.columns
```

```python
# convert Date 
df["date_of_admission"] = pd.to_datetime(
    df["date_of_admission"],
    errors="coerce"
)

df["discharge_date"] = pd.to_datetime(
    df["discharge_date"],
    errors="coerce"
)
```

```python
# lenght of stay 
df["length_of_stay"] = (
    df["discharge_date"] - df["date_of_admission"]
).dt.days 
```

```python
df["length_of_stay"].describe() 
```

```python
df[df["length_of_stay"] < 0]
```

```python
# clean categorical columns
categorical_cols = [
    "gender",
    "blood_type",
    "medical_condition",
    "insurance_provider",
    "admission_type",
    "medication",
    "test_results"
]

for col in categorical_cols:
    df[col] = df[col].astype(str).str.strip()
```

```python
print("Invalid ages:", ((df["age"] < 0) | (df["age"] > 120)).sum())
print("Invalid billing:", (df["billing_amount"] < 0).sum())
print("Invalid stay:", (df["length_of_stay"] < 0).sum())
```

```python
# create a usefull analytics colums 
print("Invalid ages:", ((df["age"] < 0) | (df["age"] > 120)).sum())
print("Invalid billing:", (df["billing_amount"] < 0).sum())
print("Invalid stay:", (df["length_of_stay"] < 0).sum())
```

```python
#AGE Grouping
df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 18, 30, 45, 60, 75, 120],
    labels=[
        "0-18",
        "19-30",
        "31-45",
        "46-60",
        "61-75",
        "76+"
    ],
    include_lowest=True
)
```

```python
# Final quality check
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum()) 
```

```python
# Export the cleaned dataset
df.to_csv(
    "../Data/healthcare_clean.csv",
    index=False
)

print("Clean dataset exported successfully.") 
```
